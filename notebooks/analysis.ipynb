week 1 Employee Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/employees_clean.csv")
df.head()
```

#salary summary by dept


```python
summary=df.groupby("department").agg(headcount=("id","count"),avg_salary=("salary","mean"),max_salary=("salary","max"))
summary
```

# speed test


```python
%timeit df.groupby("department")["salary"].mean()
```

#chart

```python
summary["avg_salary"].plot(kind="bar", title="Average salary by department", label="Average salary")
plt.xlabel("Department")
plt.ylabel("Average salary (Rs)")
plt.legend()
plt.show()
```
